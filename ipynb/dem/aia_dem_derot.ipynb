```python
import sunpy 
import sunpy.map 
from sunpy.coordinates import propagate_with_solar_surface
from glob import glob 
from tqdm.notebook import tqdm
import os 
```

```python
aia_94_files = sorted(glob("../../data/pid_1_123_aux/AIA/DEM_seq/raw/94/*.94.image.fits"))
aia_131_files = sorted(glob("../../data/pid_1_123_aux/AIA/DEM_seq/raw/131/*.131.image.fits"))
aia_171_files = sorted(glob("../../data/pid_1_123_aux/AIA/DEM_seq/raw/171/*.171.image.fits"))
aia_193_files = sorted(glob("../../data/pid_1_123_aux/AIA/DEM_seq/raw/193/*.193.image.fits"))
aia_211_files = sorted(glob("../../data/pid_1_123_aux/AIA/DEM_seq/raw/211/*.211.image.fits"))
aia_335_files = sorted(glob("../../data/pid_1_123_aux/AIA/DEM_seq/raw/335/*.335.image.fits"))
```

```python
aia_94_maps = sunpy.map.Map(aia_94_files)
```

```python
aia_94_maps_derot = [aia_94_maps[0],]
```

```python
with propagate_with_solar_surface():
    for ii in tqdm(range(1, len(aia_94_files))):
        aia_94_maps_derot.append(aia_94_maps[ii].reproject_to(aia_94_maps[0].wcs, algorithm="exact",))
```

```python
aia_94_maps_derot[150].plot()
```

```python
for ii in tqdm(range(len(aia_94_maps_derot))):
    filename_orig = os.path.basename(aia_94_files[ii]) 
    aia_94_maps_derot[ii].save(os.path.join("../../data/pid_1_123_aux/AIA/DEM_seq/derot/94/", filename_orig), overwrite=True)
```

```python
aia_94_maps_derot_seq = sunpy.map.Map(aia_94_maps_derot, sequence=True)
```

```python
anim = aia_94_maps_derot_seq.plot()
anim.save("../../figs/test_movie/aia_94_derot.mp4")
```
