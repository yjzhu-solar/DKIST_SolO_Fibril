## Generate time-distance diagrams

```python
def get_slit_data(
    x_select, y_select, line_width,
    x_select_p, y_select_p, line_width_p,
    halpha_cube=Halpha_pr_rebin[Halpha_start_index:Halpha_end_index, :, :].transpose(1, 2, 0),
    hbeta_cube=Hbeta_pr_rebin[Hbeta_start_index:Hbeta_end_index, :, :].transpose(1, 2, 0),
    CaIIK_cube=CaIIK_pr_rebin[CaIIK_start_index:CaIIK_end_index, :, :].transpose(1, 2, 0),
    hrieuv_proj_cube=hrieuv_proj_pr_da[hrieuv_start_index:hrieuv_end_index, :, :].transpose(1, 2, 0),
    aia_cube=aia171_pr_da.transpose(1, 2, 0),
    irissji_cube=irissji_pr_da[irissji_start_index:irissji_end_index, :, :].transpose(1, 2, 0),
    hrieuv_noproj_cube=eui_map_no_proj_array_crop[:, :, hrieuv_start_index:hrieuv_end_index],
    bg_kwargs={"method": "median"},
    ):

    x_select = np.asarray(x_select, dtype=float)
    y_select = np.asarray(y_select, dtype=float)

    # One slit geometry per pixel grid, reused for every cube on that grid
    slit = sip.SlitGeometry(x_select, y_select, line_width=line_width)
    slit_half = sip.SlitGeometry(x_select / 2, y_select / 2, line_width=max(line_width // 2, 1))
    slit_p = sip.SlitGeometry(x_select_p, y_select_p, line_width=line_width_p)

    def spacetime(cube, geometry):
        result = sip.extract_slits(cube, [geometry])[0]
        intensity, _ = sip.remove_background(result.intensity, **bg_kwargs)
        return result, intensity

    slit_halpha_result, slit_halpha_intensity = spacetime(halpha_cube, slit)
    _, slit_hbeta_intensity = spacetime(hbeta_cube, slit)
    _, slit_CaIIK_intensity = spacetime(CaIIK_cube, slit)

    _, slit_aia_intensity = spacetime(aia_cube, slit_half)
    _, slit_irissji_intensity = spacetime(irissji_cube, slit_half)
    _, slit_hrieuv_proj_intensity = spacetime(hrieuv_proj_cube, slit_half)

    slit_hrieuv_noproj_result, slit_hrieuv_noproj_intensity = spacetime(hrieuv_noproj_cube, slit_p)

    # .distance is the arc length along the slit in pixels (was "pixel_distance_interp")
    length_extent = slit_halpha_result.distance.value[np.r_[0, -1]] * pixel_scale
    length_extent_p = slit_hrieuv_noproj_result.distance.value[np.r_[0, -1]] * pixel_scale_p

    return ((
        slit_hbeta_intensity,
        slit_halpha_intensity,
        slit_CaIIK_intensity,
        slit_irissji_intensity,
        slit_aia_intensity,
        slit_hrieuv_noproj_intensity,
        slit_hrieuv_proj_intensity,
    ),
    (length_extent, length_extent_p))
```

```python
def plot_slit_data(
    slit_data,
    axes,
    length_extent,
    time_extents=[
        Hbeta_time_extent,
        Halpha_time_extent,
        CaIIK_time_extent,
        irissji_time_extent,
        aia_time_extent,
        hrieuv_time_extent
    ],
    norms=[None, None, None, None, None],
    ):

    cmap = ("Greys_r", "Greys_r", "bone", "irissji1400", "sdoaia171", "sdoaia171", "sdoaia171")

    for ii in range(6):
        axes[ii].imshow(
        slit_data[ii], cmap=cmap[ii], origin="lower", interpolation="none",
        norm=norms[ii],
        aspect="auto",
        extent=[*time_extents[ii], *length_extent[0]],
    )

    axes[6].imshow(
        slit_data[6], cmap=cmap[6], origin="lower", interpolation="none",
        norm=norms[6],
        aspect="auto",
        extent=[*time_extents[5], *length_extent[1]],
    )

    return None
```

```python
def percentile_norm(vmin, vmax, data, asinh=0.4):
    if asinh is None:
        return ImageNormalize(
            vmin=np.nanpercentile(data, vmin),
            vmax=np.nanpercentile(data, vmax),
        )
    else:
        return ImageNormalize(
            vmin=np.nanpercentile(data, vmin),
            vmax=np.nanpercentile(data, vmax),
            stretch=AsinhStretch(asinh),
        )
```
